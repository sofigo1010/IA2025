## 1. ¿Qué es el Temporal Difference Learning y en qué se diferencia de los métodos tradicionales de aprendizaje supervisado?

**Temporal Difference Learning (TD Learning):**  
Es un método de aprendizaje por refuerzo que permite actualizar de forma incremental la estimación de la función de valor a partir de la diferencia entre predicciones sucesivas. En lugar de esperar a la finalización de un episodio (como en los métodos Monte Carlo), TD Learning utiliza la retroalimentación inmediata obtenida en cada paso para ajustar las predicciones.

**Diferencias con el aprendizaje supervisado:**  
- **Etiquetado de datos:**  
  En el aprendizaje supervisado se requiere un conjunto de datos etiquetado (entrada-salida) para entrenar el modelo, mientras que en TD Learning no se cuentan con “etiquetas” explícitas; el aprendizaje se basa en la experiencia interactiva con el entorno.
- **Retroalimentación y actualización:**  
  En métodos supervisados, el ajuste se realiza comparando la predicción con una etiqueta verdadera. En TD Learning, la "etiqueta" se genera combinando la recompensa inmediata con la estimación del siguiente estado, utilizando lo que se conoce como **error de diferencia temporal**.

**Error de Diferencia Temporal (TD Error):**  
Se define como:
  
$$
\delta = r + \gamma V(s') - V(s)
$$

donde:  
- $`r`$ es la recompensa recibida,  
- $`\gamma`$ es el factor de descuento,  
- $`V(s)`$ es la estimación del valor del estado actual,  
- $`V(s')`$ es la estimación del valor del siguiente estado.

El TD error mide la discrepancia entre la predicción actual y la nueva estimación basada en la experiencia inmediata, y se utiliza para actualizar la función de valor de forma continua.

## 2. Decisiones en Juegos Simultáneos

En los **juegos simultáneos**, los jugadores toman decisiones al mismo tiempo, sin conocer las acciones que elegirán sus oponentes. Debido a esta incertidumbre, cada jugador debe basar su estrategia en las expectativas sobre el comportamiento de los demás.

**Ejemplo del mundo real:**  
- **Subastas cerradas:**  
  En una subasta cerrada, todos los participantes presentan sus ofertas simultáneamente sin conocer las ofertas de los demás. Cada postor debe estimar el valor del objeto y anticipar las pujas de sus competidores para decidir su propia oferta.

**Estrategias posibles:**  
- **Estrategia mixta:** Los jugadores pueden asignar probabilidades a diferentes acciones, mezclando sus estrategias para evitar ser predecibles.  
- **Modelado de creencias:** Cada jugador forma hipótesis sobre las posibles acciones de sus oponentes basándose en información histórica o en el contexto del problema.  
- **Optimización bajo incertidumbre:** Se busca maximizar la recompensa esperada, considerando los posibles resultados derivados de la incertidumbre en las acciones adversarias.

---

## 3. Juegos de Suma Cero vs. Juegos de No Suma Cero

**Juegos de Suma Cero:**  
- En estos juegos, la ganancia de un jugador es exactamente la pérdida del otro, de modo que el total de beneficios y pérdidas suma cero.  
- **Implicación en la toma de decisiones:**  
  Cada ventaja obtenida por un jugador implica una desventaja equivalente para su oponente, lo que fomenta estrategias estrictamente competitivas.

**Juegos de No Suma Cero:**  
- En estos juegos, el resultado total no es necesariamente cero, pudiendo existir situaciones en las que ambos jugadores ganen o pierdan simultáneamente.
- **Ejemplo:**  
  El **dilema del prisionero** es un claro ejemplo, donde la tentación individual de traicionar puede ser alta, pero la cooperación mutua produce un resultado mejor para ambos.
- **Consideraciones estratégicas:**  
  - **Cooperación vs. competencia:** Existe la posibilidad de estrategias cooperativas que beneficien a ambas partes.  
  - **Soluciones Pareto eficientes:** Los jugadores pueden buscar acuerdos o equilibrios en los que ninguna mejora unilateral sea posible sin perjudicar a la otra parte.  
  - **Equilibrio de Nash en contextos cooperativos:** La estabilidad del equilibrio puede requerir compromisos y mecanismos para fomentar la confianza entre los jugadores.

---

## 4. Equilibrio de Nash en Juegos Simultáneos

El **equilibrio de Nash** se aplica en juegos simultáneos como una situación en la que cada jugador elige una estrategia de modo que, dado el conjunto de estrategias de los demás, ningún jugador tiene incentivos para cambiar su elección de manera unilateral.

- **Estabilidad:**  
  En el equilibrio de Nash, cualquier cambio individual sin coordinación llevaría a un peor resultado para el jugador que se desvía.
- **Aplicación en juegos simultáneos:**  
  Dado que los jugadores seleccionan sus estrategias sin conocer las elecciones de los demás, el equilibrio de Nash es una predicción razonable de comportamiento en escenarios con información incompleta.
- **Ejemplo:**  
  En el dilema del prisionero, si ambos jugadores adoptan la estrategia que representa el equilibrio de Nash (generalmente la traición en el contexto clásico), ninguno ganará en mejorar su resultado cambiando de forma unilateral.

## 5. Aplicación del Temporal Difference Learning en Procesos de Toma de Decisiones Dinámicos

**Modelado y Optimización en Entornos Dinámicos:**  
TD Learning se utiliza para optimizar procesos de toma de decisiones en entornos donde las condiciones cambian rápidamente. Al actualizar continuamente las estimaciones de valor en función de la experiencia inmediata, el método se adapta dinámicamente a nuevos escenarios y recompensas.

**Equilibrio entre Exploración y Explotación:**  
- **Explotación:** Se aprovecha el conocimiento actual para maximizar la recompensa inmediata.  
- **Exploración:** Se prueban nuevas acciones para descubrir información que pueda conducir a mejores recompensas en el futuro.
- **Mecanismos comunes:**  
  - Estrategia **$\varepsilon$-greedy**, en la que con probabilidad $`\varepsilon`$ se elige una acción aleatoria (exploración) y con probabilidad $`1-\varepsilon`$ se selecciona la acción óptima (explotación).  
  - Otros enfoques incluyen métodos basados en la incertidumbre, como el Upper Confidence Bound (UCB).

**Desafíos en la Implementación Práctica:**  
- **Convergencia:** Asegurar que las actualizaciones basadas en el TD error conduzcan a una función de valor estable.  
- **Selección del parámetro de exploración:** Determinar la tasa de exploración adecuada para evitar tanto la exploración excesiva como la explotación prematura.  
- **Aproximación de funciones:** En entornos con un gran número de estados o estados continuos, se utilizan métodos de aproximación (como redes neuronales), lo que puede introducir problemas de estabilidad y convergencia.  
- **Equilibrio a largo plazo:** Manejar la discrepancia entre la recompensa inmediata y la recompensa futura, crucial en entornos complejos y no estacionarios.




# Laboratorio 7 – Connect Four con TD Learning

En este notebook se implementa un agente para Connect Four basado en Temporal Difference Learning (TD Learning).  
El agente aprende a partir de sus propias experiencias actualizando una función de valor basada en el error de diferencia temporal.  

Los pasos principales son:
1. **Representación del Estado:**  
   Se utiliza el tablero (una matriz de NumPy) y se codifica en un vector (o tupla) que sirva de clave para la tabla Q.
2. **Espacio de Acción:**  
   Las acciones corresponden a las columnas en las que se puede dejar caer una pieza, es decir, las posiciones válidas del tablero.
3. **Algoritmo TD (Q-learning):**  
   La actualización de la función de valor se realiza usando la regla de Q-learning:  
   $$ Q(s,a) \leftarrow Q(s,a) + \alpha \left[ r + \gamma \max_{a'} Q(s',a') - Q(s,a) \right] $$
4. **Definición de Recompensas:**  
   - Victoria del agente: recompensa +1  
   - Derrota: recompensa -1  
   - Empate: 0  
   También se pueden definir recompensas intermedias para incentivar comportamientos favorables.
5. **Estrategia de Exploración:**  
   Se emplea una política $\varepsilon$-greedy para equilibrar exploración y explotación.
6. **Ciclo de Entrenamiento y Evaluación:**  
   El agente TD se enfrenta a dos tipos de agentes Minimax (uno sin poda y otro con poda alfa-beta) en 50 juegos cada uno, totalizando 150 juegos.  
   Luego se grafica la cantidad de victorias de cada agente.

```python
import numpy as np
import pygame
import sys
import math
import random
import time

# Parámetros del juego
ROWS = 6
COLS = 7
EMPTY = 0

PLAYER = 1   # Humano (rojo)
AI_1 = 2     # Agente TD (Amarillo)
AI_2 = 3     # Agente Minimax (Naranja)

pygame.init()
WIDTH = 700
HEIGHT = 700
SQUARESIZE = WIDTH // COLS
RADIUS = SQUARESIZE // 2 - 5
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Connect 4 - TD Learning vs Minimax")

FONT_TITLE = pygame.font.SysFont("monospace", 35, bold=True)
FONT_OPTION = pygame.font.SysFont("monospace", 30)
FONT_WINNER = pygame.font.SysFont("monospace", 50, bold=True)
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 200, 0)
BLUE = (0, 0, 255)
RED = (255, 0, 0)
YELLOW = (255, 255, 0)
ORANGE = (255, 165, 0)

def create_board():
    return np.zeros((ROWS, COLS), dtype=int)

def drop_piece(board, col, piece):
    row = get_next_open_row(board, col)
    if row is not None:
        board[row][col] = piece
        return True
    return False

def get_next_open_row(board, col):
    for r in range(ROWS - 1, -1, -1):
        if board[r][col] == EMPTY:
            return r
    return None

def is_valid_location(board, col):
    return board[0][col] == EMPTY

def get_valid_locations(board):
    return [c for c in range(COLS) if is_valid_location(board, c)]

def winning_move(board, piece):
    # Chequeo horizontal
    for r in range(ROWS):
        for c in range(COLS - 3):
            if all(board[r, c + i] == piece for i in range(4)):
                return True
    # Chequeo vertical
    for c in range(COLS):
        for r in range(ROWS - 3):
            if all(board[r + i, c] == piece for i in range(4)):
                return True
    # Chequeo diagonal
    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            if all(board[r + i, c + i] == piece for i in range(4)):
                return True
            if all(board[r + 3 - i, c + i] == piece for i in range(4)):
                return True
    return False

def board_full(board):
    return len(get_valid_locations(board)) == 0

def draw_board(board):
    screen.fill(BLUE)
    # Dibuja el tablero vacío
    for c in range(COLS):
        for r in range(ROWS):
            pygame.draw.circle(screen, BLACK, (c * SQUARESIZE + SQUARESIZE // 2, (r + 1) * SQUARESIZE), RADIUS)
    # Dibuja las piezas
    for c in range(COLS):
        for r in range(ROWS):
            if board[r][c] == PLAYER:
                pygame.draw.circle(screen, RED, (c * SQUARESIZE + SQUARESIZE // 2, HEIGHT - (ROWS - r) * SQUARESIZE), RADIUS)
            elif board[r][c] == AI_1:
                pygame.draw.circle(screen, YELLOW, (c * SQUARESIZE + SQUARESIZE // 2, HEIGHT - (ROWS - r) * SQUARESIZE), RADIUS)
            elif board[r][c] == AI_2:
                pygame.draw.circle(screen, ORANGE, (c * SQUARESIZE + SQUARESIZE // 2, HEIGHT - (ROWS - r) * SQUARESIZE), RADIUS)
    pygame.display.update()

def show_winner(message):
    screen.fill(BLACK)
    text = FONT_WINNER.render(message, True, WHITE)
    text_rect = text.get_rect(center=(WIDTH // 2, HEIGHT // 2))
    screen.blit(text, text_rect)
    pygame.display.update()
    pygame.time.wait(3000)

def draw_menu(title, options, selected):
    screen.fill(BLACK)
    title_label = FONT_TITLE.render(title, True, WHITE)
    title_rect = title_label.get_rect(center=(WIDTH // 2, 100))
    screen.blit(title_label, title_rect)
    for i, option in enumerate(options):
        color = GREEN if i == selected else WHITE
        label = FONT_OPTION.render(option, True, color)
        label_rect = label.get_rect(center=(WIDTH // 2, 200 + i * 80))
        screen.blit(label, label_rect)
    pygame.display.update()

def menu_selection(title, options):
    selected = 0
    while True:
        draw_menu(title, options, selected)
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            elif event.type == pygame.KEYDOWN:
                if event.key == pygame.K_DOWN:
                    selected = (selected + 1) % len(options)
                elif event.key == pygame.K_UP:
                    selected = (selected - 1) % len(options)
                elif event.key == pygame.K_RETURN:
                    return selected

def evaluate_window(window):
    score = 0
    count_AI1 = window.count(AI_1)
    count_AI2 = window.count(AI_2)
    count_empty = window.count(EMPTY)
    if count_AI1 == 4:
        score += 10000
    elif count_AI1 == 3 and count_empty == 1:
        score += 100
    elif count_AI1 == 2 and count_empty == 2:
        score += 10
    if count_AI2 == 4:
        score -= 10000
    elif count_AI2 == 3 and count_empty == 1:
        score -= 100
    elif count_AI2 == 2 and count_empty == 2:
        score -= 10
    return score

def evaluate_board(board):
    score = 0
    center = COLS // 2
    center_array = [board[r][center] for r in range(ROWS)]
    center_count_AI1 = center_array.count(AI_1)
    score += center_count_AI1 * 6
    for r in range(ROWS):
        row_array = board[r, :]
        for c in range(COLS - 3):
            window = list(row_array[c : c+4])
            score += evaluate_window(window)
    for c in range(COLS):
        col_array = board[:, c]
        for r in range(ROWS - 3):
            window = list(col_array[r : r+4])
            score += evaluate_window(window)
    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            window_down = [board[r+i, c+i] for i in range(4)]
            score += evaluate_window(window_down)
            window_up = [board[r+3-i, c+i] for i in range(4)]
            score += evaluate_window(window_up)
    return score

def minimax(board, depth, alpha, beta, maximizing_player, use_alpha_beta):
    valid_cols = get_valid_locations(board)
    game_over = (winning_move(board, AI_1) or winning_move(board, AI_2) or len(valid_cols) == 0)
    if depth == 0 or game_over:
        return None, evaluate_board(board)
    if maximizing_player:
        value = -math.inf
        best_col = random.choice(valid_cols)
        for col in valid_cols:
            temp = board.copy()
            drop_piece(temp, col, AI_1)
            _, new_score = minimax(temp, depth-1, alpha, beta, False, use_alpha_beta)
            if new_score > value:
                value = new_score
                best_col = col
            if use_alpha_beta:
                alpha = max(alpha, value)
                if alpha >= beta:
                    break
        return best_col, value
    else:
        value = math.inf
        best_col = random.choice(valid_cols)
        for col in valid_cols:
            temp = board.copy()
            drop_piece(temp, col, AI_2)
            _, new_score = minimax(temp, depth-1, alpha, beta, True, use_alpha_beta)
            if new_score < value:
                value = new_score
                best_col = col
            if use_alpha_beta:
                beta = min(beta, value)
                if alpha >= beta:
                    break
        return best_col, value

def get_immediate_move(board, piece, rival):
    valid_cols = get_valid_locations(board)
    for col in valid_cols:
        temp_board = board.copy()
        drop_piece(temp_board, col, piece)
        if winning_move(temp_board, piece):
            return col
    for col in valid_cols:
        temp_board = board.copy()
        drop_piece(temp_board, col, rival)
        if winning_move(temp_board, rival):
            return col
    return None

def get_best_move(board, piece, rival, depth, use_alpha_beta):
    col_urgent = get_immediate_move(board, piece, rival)
    if col_urgent is not None:
        return col_urgent
    if piece == AI_1:
        col, _ = minimax(board, depth, -math.inf, math.inf, True, use_alpha_beta)
    else:
        col, _ = minimax(board, depth, -math.inf, math.inf, False, use_alpha_beta)
    return col

# ------------------ TD LEARNING CON Q-LEARNING ------------------

class TDAgent:
    def __init__(self, alpha=0.1, gamma=0.9, epsilon=0.1):
        self.alpha = alpha       # Tasa de aprendizaje
        self.gamma = gamma       # Factor de descuento
        self.epsilon = epsilon   # Tasa de exploración
        self.q_table = {}        # Diccionario: estado -> {acción: valor Q}

    def get_state(self, board):
        # Se convierte el tablero en una tupla para usarlo como clave
        return tuple(board.flatten())

    def choose_action(self, board, valid_actions):
        state = self.get_state(board)
        if state not in self.q_table:
            self.q_table[state] = {a: 0 for a in valid_actions}
        # Política epsilon-greedy
        if random.random() < self.epsilon:
            return random.choice(valid_actions)
        else:
            q_values = self.q_table[state]
            max_q = max(q_values[a] for a in valid_actions)
            best_actions = [a for a in valid_actions if q_values[a] == max_q]
            return random.choice(best_actions)

    def update(self, board, action, reward, next_board, done, valid_actions_next):
        state = self.get_state(board)
        next_state = self.get_state(next_board)
        if state not in self.q_table:
            self.q_table[state] = {a: 0 for a in valid_actions_next}
        if next_state not in self.q_table:
            self.q_table[next_state] = {a: 0 for a in valid_actions_next}
        current_q = self.q_table[state].get(action, 0)
        if done:
            max_next_q = 0
        else:
            max_next_q = max(self.q_table[next_state].get(a, 0) for a in valid_actions_next)
        # Actualización según la regla de Q-learning
        self.q_table[state][action] = current_q + self.alpha * (reward + self.gamma * max_next_q - current_q)

# ------------------------------------------------------------------

# Juego entre el agente TD (AI_1) y el agente Minimax (AI_2)
def play_game_td_vs_minimax(use_alpha_beta):
    board = create_board()
    game_over = False
    current_player = AI_1  # El agente TD inicia
    # Inicializamos el agente TD
    td_agent = TDAgent(alpha=0.1, gamma=0.9, epsilon=0.2)
    state_old = board.copy()  # Para la actualización de TD
    while not game_over:
        draw_board(board)
        time.sleep(0.3)
        valid_moves = get_valid_locations(board)
        if current_player == AI_1:
            # Turno del agente TD
            action = td_agent.choose_action(board, valid_moves)
            drop_piece(board, action, AI_1)
            if winning_move(board, AI_1):
                reward = 1
                show_winner("¡TD Agent (Amarillo) ganó!")
                game_over = True
            elif board_full(board):
                reward = 0
                show_winner("¡Empate!")
                game_over = True
            else:
                reward = 0
            next_valid = get_valid_locations(board)
            td_agent.update(state_old, action, reward, board.copy(), game_over, next_valid)
            state_old = board.copy()
            current_player = AI_2
        else:
            # Turno del agente Minimax
            depth = 4 if not use_alpha_beta else 6
            col = get_best_move(board, AI_2, AI_1, depth, use_alpha_beta)
            drop_piece(board, col, AI_2)
            if winning_move(board, AI_2):
                show_winner("Minimax (Naranja) ganó")
                game_over = True
            elif board_full(board):
                show_winner("¡Empate!")
                game_over = True
            current_player = AI_1
    if menu_selection("¿Jugar de nuevo?", ["Sí", "No"]) == 0:
        main()
    else:
        pygame.quit()
        sys.exit()

def play_game_human_vs_ai(use_alpha_beta):
    board = create_board()
    game_over = False
    turn = 0  
    depth_if_alpha = 6
    depth_if_noalpha = 4
    while not game_over:
        draw_board(board)
        if turn == 0:
            event = pygame.event.wait()
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.MOUSEBUTTONDOWN:
                x_pos = event.pos[0]
                col = x_pos // SQUARESIZE
                if is_valid_location(board, col):
                    drop_piece(board, col, PLAYER)
                    if winning_move(board, PLAYER):
                        show_winner("¡Ganaste (Humano)!")
                        game_over = True
                    else:
                        if board_full(board):
                            show_winner("¡Empate!")
                            game_over = True
                    turn = 1
        else:
            time.sleep(0.3)
            depth = depth_if_alpha if use_alpha_beta else depth_if_noalpha
            col = get_best_move(board, AI_1, PLAYER, depth, use_alpha_beta)
            drop_piece(board, col, AI_1)
            if winning_move(board, AI_1):
                show_winner("La IA (Amarillo) ganó")
                game_over = True
            else:
                if board_full(board):
                    show_winner("¡Empate!")
                    game_over = True
            turn = 0
    if menu_selection("¿Jugar de nuevo?", ["Sí", "No"]) == 0:
        main()
    else:
        pygame.quit()
        sys.exit()

def play_game_ai_vs_ai(alpha_beta_ai):
    board = create_board()
    game_over = False
    depth_alpha = 6
    depth_noalpha = 4
    current_player = random.choice([AI_1, AI_2])
    first_move_done = False
    while not game_over:
        draw_board(board)
        time.sleep(0.3)
        use_poda = (alpha_beta_ai == current_player)
        depth = depth_alpha if use_poda else depth_noalpha
        rival = AI_2 if current_player == AI_1 else AI_1
        if not first_move_done:
            valid_cols = get_valid_locations(board)
            col = random.choice(valid_cols)
            first_move_done = True
        else:
            col = get_best_move(board, current_player, rival, depth, use_poda)
        drop_piece(board, col, current_player)
        if winning_move(board, current_player):
            if current_player == AI_1:
                show_winner("IA_1 (Amarillo) ganó")
            else:
                show_winner("IA_2 (Naranja) ganó")
            game_over = True
        else:
            if board_full(board):
                show_winner("¡Empate!")
                game_over = True
        current_player = AI_2 if current_player == AI_1 else AI_1
    if menu_selection("¿Jugar de nuevo?", ["Sí", "No"]) == 0:
        main()
    else:
        pygame.quit()
        sys.exit()

def main():
    mode = menu_selection("Modo de juego", ["Humano vs IA", "IA vs IA", "TD vs Minimax"])
    if mode == 0:
        use_pruning = (menu_selection("¿Activar poda Alfa-Beta (IA)?", ["Sí", "No"]) == 0)
        play_game_human_vs_ai(use_pruning)
    elif mode == 1:
        who_uses_pruning = menu_selection("¿Quién usa la poda Alfa-Beta?", ["AI_1 (Amarillo)", "AI_2 (Naranja)", "Ninguno"])
        if who_uses_pruning == 0:
            alpha_beta_ai = AI_1
        elif who_uses_pruning == 1:
            alpha_beta_ai = AI_2
        else:
            alpha_beta_ai = None
        play_game_ai_vs_ai(alpha_beta_ai)
    else:
        use_pruning = (menu_selection("¿Activar poda Alfa-Beta para Minimax (Naranja)?", ["Sí", "No"]) == 0)
        play_game_td_vs_minimax(use_pruning)

if __name__ == "__main__":
    main()
```
