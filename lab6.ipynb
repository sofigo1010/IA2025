### **1. En un juego de suma cero para dos jugadores, ¿cómo funciona el algoritmo minimax para determinar la estrategia óptima para cada jugador? ¿Puede explicarnos el concepto de "valor minimax" y su importancia en este contexto?**

El **algoritmo Minimax** es una técnica utilizada en juegos de suma cero para dos jugadores con información perfecta, como el ajedrez o el tres en línea. Su objetivo es determinar la mejor estrategia posible para un jugador, asumiendo que el oponente también juega de manera óptima.

### **Funcionamiento del Algoritmo Minimax**
1. **Construcción del Árbol de Juego:**  
   - Se generan todas las posibles jugadas hasta una cierta profundidad.
   - Cada nodo representa un estado del juego.
   - Los nodos terminales tienen valores asociados que indican qué tan favorables son para el jugador MAX.

2. **Asignación de Valores a los Estados Terminales:**  
   - Se asignan valores numéricos a los estados finales desde la perspectiva del jugador MAX:
     - **Valor positivo:** Ventaja para MAX.
     - **Valor negativo:** Ventaja para MIN.
     - **Valor cero:** Empate.

3. **Propagación de los Valores Minimax:**  
   - Se evalúan los nodos desde las hojas hasta la raíz:
     - En los turnos de **MAX**, se elige la acción que maximiza la ganancia.
     - En los turnos de **MIN**, se elige la acción que minimiza la ganancia de MAX.

4. **Selección de la Mejor Jugada:**  
   - Al llegar a la raíz, el algoritmo selecciona la jugada con el mejor valor para MAX, garantizando la mejor estrategia si el oponente juega de manera óptima.

### **Concepto de "Valor Minimax"**
El **valor minimax** de un nodo representa la mejor puntuación que un jugador puede garantizarse, suponiendo que su oponente también juega de la mejor manera posible. Se obtiene mediante la propagación de valores en el árbol:

- **Si es el turno de MAX**, elige el valor máximo entre sus opciones.
- **Si es el turno de MIN**, elige el valor mínimo entre sus opciones.

Este valor es crucial porque permite tomar decisiones óptimas bajo la suposición de que el oponente también jugará de forma óptima.

### **Importancia del Valor Minimax**
1. **Garantiza la mejor estrategia posible**: Permite que un jugador juegue de la manera más óptima contra un oponente racional.
2. **Aplicable a múltiples juegos**: Se utiliza en ajedrez, damas, tres en línea y otros juegos de estrategia.
3. **Forma la base de algoritmos avanzados**: Técnicas como **Alpha-Beta Pruning** optimizan Minimax reduciendo la cantidad de nodos evaluados.


### **2. Compare y contraste el algoritmo minimax con la poda alfa-beta. ¿Cómo mejora la poda alfa-beta la eficiencia del algoritmo minimax, particularmente en árboles de caza grandes? Proporcione un ejemplo para ilustrar la diferencia en la complejidad computacional entre la poda minimax y alfa-beta**

El **algoritmo Minimax** y la **poda Alfa-Beta** son estrategias utilizadas en juegos de dos jugadores para encontrar la mejor jugada. Sin embargo, la poda Alfa-Beta es una mejora del Minimax que lo hace más eficiente al reducir la cantidad de nodos evaluados.

### **Diferencias Clave entre Minimax y Poda Alfa-Beta**
| **Característica**   | **Minimax** | **Poda Alfa-Beta** |
|---------------------|------------|-------------------|
| **Búsqueda**       | Evalúa todos los nodos posibles. | Descarta nodos innecesarios. |
| **Eficiencia**     | Lento en árboles grandes. | Más rápido al eliminar ramas inútiles. |
| **Complejidad**    | \(O(b^d)\) (donde \(b\) es el número de movimientos y \(d\) la profundidad). | \(O(b^{d/2})\) en el mejor caso. |
| **Uso de memoria** | Alto, ya que explora todo el árbol. | Menor, porque evita cálculos innecesarios. |

### **¿Cómo mejora la poda Alfa-Beta la eficiencia de Minimax?**
La poda Alfa-Beta **reduce el número de nodos evaluados** al eliminar ramas que no pueden afectar la decisión final. Funciona manteniendo dos valores:

- **Alfa (α):** El valor máximo que el jugador MAX puede asegurar.
- **Beta (β):** El valor mínimo que el jugador MIN puede asegurar.

Si un nodo genera un valor peor que el límite actual de **α o β**, entonces se **descarta** porque no influirá en la decisión final.

### **Ejemplo de Diferencia en Complejidad**
Supongamos un juego donde cada jugador tiene 3 opciones por turno y analizamos hasta 4 niveles de profundidad:

- **Minimax puro:** Examina **\(3^4 = 81\) nodos**.
- **Poda Alfa-Beta (mejor caso):** Puede reducir la búsqueda hasta **\(3^2 = 9\) nodos**.

Esto muestra cómo Alfa-Beta **reduce drásticamente los cálculos**, acelerando la toma de decisiones sin perder precisión en la estrategia.

### **Conclusión**
La poda Alfa-Beta es simplemente una versión optimizada del Minimax que **hace lo mismo pero más rápido**, descartando movimientos que no afectan la jugada óptima. Es crucial en juegos como el ajedrez, donde la cantidad de posibles movimientos crece exponencialmente.


### **3. ¿Cuál es el papel de expectiminimax en juegos con incertidumbre, como aquellos que involucran nodos de azar o información oculta? ¿En qué se diferencia el expectiminimax del minimax en el manejo de resultados probabilísticos y cuáles son los desafíos clave que aborda?**

El **algoritmo Expectiminimax** es una extensión del **Minimax**, diseñado para juegos donde hay **incertidumbre**, como los que incluyen **dados, cartas ocultas o eventos aleatorios**. Se utiliza en juegos como **Backgammon**, donde el resultado de una acción depende del azar.

### **¿Cómo Funciona Expectiminimax?**
El árbol de decisión en Expectiminimax tiene **tres tipos de nodos**:

1. **Nodos MAX**: Representan el turno del jugador que intenta **maximizar** su ganancia.
2. **Nodos MIN**: Representan el turno del oponente, que intenta **minimizar** la ganancia del jugador MAX.
3. **Nodos de Azar**: Representan eventos aleatorios (como lanzar un dado o robar una carta). En estos nodos, el valor se calcula como un **promedio ponderado** de los posibles resultados según sus probabilidades.

### **Diferencias Clave entre Minimax y Expectiminimax**
| **Característica** | **Minimax** | **Expectiminimax** |
|-------------------|------------|-------------------|
| **Tipo de juego** | Juegos deterministas (ajedrez, damas). | Juegos con azar (Backgammon, Póker). |
| **Evaluación de nodos** | Se elige el valor máximo o mínimo en cada turno. | Se calcula el valor esperado basado en probabilidades. |
| **Estrategia** | Asume que el oponente juega de manera óptima. | Considera resultados aleatorios y probabilidades. |

### **Desafíos que Aborda Expectiminimax**
1. **Manejo de Incertidumbre**: Permite planificar estrategias en juegos donde el resultado no depende solo de las decisiones de los jugadores, sino también del azar.
2. **Crecimiento del Árbol**: El árbol de búsqueda es más grande porque incluye nodos de probabilidad, lo que aumenta la complejidad computacional.
3. **Cálculo de Probabilidades**: Requiere conocer y usar distribuciones de probabilidad correctas para evaluar decisiones de manera precisa.

### **Ejemplo de Expectiminimax**
Supongamos un juego con un dado de 6 caras donde un jugador decide entre:
- **Opción A**: Gana **10 puntos** si el dado saca un número par.
- **Opción B**: Gana **7 puntos** si el dado saca un número impar.

El cálculo del valor esperado sería:

$$
V(A) = \left(\frac{3}{6}\right) \times 10 = 5
$$

$$
V(B) = \left(\frac{3}{6}\right) \times 7 = 3.5
$$

Dado que **5 > 3.5**, el jugador elegiría **Opción A**.

```python
import numpy as np
import pygame
import sys
import math
import random
import time
ROWS = 6
COLS = 7
EMPTY = 0

PLAYER = 1   # Humano (rojo)
AI_1 = 2     # Amarillo (MAX)
AI_2 = 3     # Naranja (MIN)

pygame.init()
WIDTH = 700
HEIGHT = 700
SQUARESIZE = WIDTH // COLS
RADIUS = SQUARESIZE // 2 - 5
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Connect 4 - Minimax con o sin Alfa-Beta")

FONT_TITLE = pygame.font.SysFont("monospace", 35, bold=True)
FONT_OPTION = pygame.font.SysFont("monospace", 30)
FONT_WINNER = pygame.font.SysFont("monospace", 50, bold=True)
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 200, 0)
BLUE = (0, 0, 255)
RED = (255, 0, 0)
YELLOW = (255, 255, 0)
ORANGE = (255, 165, 0)

def create_board():
    return np.zeros((ROWS, COLS), dtype=int)

def drop_piece(board, col, piece):

    row = get_next_open_row(board, col)
    if row is not None:
        board[row][col] = piece
        return True
    return False

def get_next_open_row(board, col):
    for r in range(ROWS - 1, -1, -1):
        if board[r][col] == EMPTY:
            return r
    return None

def is_valid_location(board, col):
    return board[0][col] == EMPTY

def get_valid_locations(board):
    return [c for c in range(COLS) if is_valid_location(board, c)]

def winning_move(board, piece):
    for r in range(ROWS):
        for c in range(COLS - 3):
            if all(board[r, c + i] == piece for i in range(4)):
                return True
    for c in range(COLS):
        for r in range(ROWS - 3):
            if all(board[r + i, c] == piece for i in range(4)):
                return True
    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            if all(board[r + i, c + i] == piece for i in range(4)):
                return True
            if all(board[r + 3 - i, c + i] == piece for i in range(4)):
                return True
    return False

def board_full(board):
    return len(get_valid_locations(board)) == 0


def draw_board(board):
    screen.fill(BLUE)
    for c in range(COLS):
        for r in range(ROWS):
            pygame.draw.circle(
                screen, BLACK,
                (c * SQUARESIZE + SQUARESIZE // 2, (r + 1) * SQUARESIZE),
                RADIUS
            )
    for c in range(COLS):
        for r in range(ROWS):
            if board[r][c] == PLAYER:
                pygame.draw.circle(
                    screen, RED,
                    (c * SQUARESIZE + SQUARESIZE // 2,
                     HEIGHT - (ROWS - r) * SQUARESIZE),
                    RADIUS
                )
            elif board[r][c] == AI_1:
                pygame.draw.circle(
                    screen, YELLOW,
                    (c * SQUARESIZE + SQUARESIZE // 2,
                     HEIGHT - (ROWS - r) * SQUARESIZE),
                    RADIUS
                )
            elif board[r][c] == AI_2:
                pygame.draw.circle(
                    screen, ORANGE,
                    (c * SQUARESIZE + SQUARESIZE // 2,
                     HEIGHT - (ROWS - r) * SQUARESIZE),
                    RADIUS
                )
    pygame.display.update()

def show_winner(message):
    screen.fill(BLACK)
    text = FONT_WINNER.render(message, True, WHITE)
    text_rect = text.get_rect(center=(WIDTH // 2, HEIGHT // 2))
    screen.blit(text, text_rect)
    pygame.display.update()
    pygame.time.wait(3000)

def draw_menu(title, options, selected):
    screen.fill(BLACK)
    title_label = FONT_TITLE.render(title, True, WHITE)
    title_rect = title_label.get_rect(center=(WIDTH // 2, 100))
    screen.blit(title_label, title_rect)
    
    for i, option in enumerate(options):
        color = GREEN if i == selected else WHITE
        label = FONT_OPTION.render(option, True, color)
        label_rect = label.get_rect(center=(WIDTH // 2, 200 + i * 80))
        screen.blit(label, label_rect)

    pygame.display.update()

def menu_selection(title, options):
    selected = 0
    while True:
        draw_menu(title, options, selected)
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            elif event.type == pygame.KEYDOWN:
                if event.key == pygame.K_DOWN:
                    selected = (selected + 1) % len(options)
                elif event.key == pygame.K_UP:
                    selected = (selected - 1) % len(options)
                elif event.key == pygame.K_RETURN:
                    return selected


def evaluate_window(window):
    score = 0
    count_AI1 = window.count(AI_1)
    count_AI2 = window.count(AI_2)
    count_empty = window.count(EMPTY)


    if count_AI1 == 4:
        score += 10000
    elif count_AI1 == 3 and count_empty == 1:
        score += 100
    elif count_AI1 == 2 and count_empty == 2:
        score += 10

    if count_AI2 == 4:
        score -= 10000
    elif count_AI2 == 3 and count_empty == 1:
        score -= 100
    elif count_AI2 == 2 and count_empty == 2:
        score -= 10

    return score

def evaluate_board(board):

    score = 0

    center = COLS // 2
    center_array = [board[r][center] for r in range(ROWS)]
    center_count_AI1 = center_array.count(AI_1)
    score += center_count_AI1 * 6


    for r in range(ROWS):
        row_array = board[r, :]
        for c in range(COLS - 3):
            window = list(row_array[c : c+4])
            score += evaluate_window(window)

    for c in range(COLS):
        col_array = board[:, c]
        for r in range(ROWS - 3):
            window = list(col_array[r : r+4])
            score += evaluate_window(window)

    for r in range(ROWS - 3):
        for c in range(COLS - 3):
            window_down = [board[r+i, c+i] for i in range(4)]
            score += evaluate_window(window_down)
            window_up = [board[r+3-i, c+i] for i in range(4)]
            score += evaluate_window(window_up)

    return score

def minimax(board, depth, alpha, beta, maximizing_player, use_alpha_beta):
    valid_cols = get_valid_locations(board)
    game_over = (winning_move(board, AI_1) or 
                 winning_move(board, AI_2) or 
                 len(valid_cols) == 0)

    if depth == 0 or game_over:
        return None, evaluate_board(board)

    if maximizing_player:
        value = -math.inf
        best_col = random.choice(valid_cols)
        for col in valid_cols:
            temp = board.copy()
            drop_piece(temp, col, AI_1)
            _, new_score = minimax(temp, depth-1, alpha, beta, False, use_alpha_beta)
            if new_score > value:
                value = new_score
                best_col = col
            if use_alpha_beta:
                alpha = max(alpha, value)
                if alpha >= beta:
                    break
        return best_col, value
    else:

        value = math.inf
        best_col = random.choice(valid_cols)
        for col in valid_cols:
            temp = board.copy()
            drop_piece(temp, col, AI_2)
            _, new_score = minimax(temp, depth-1, alpha, beta, True, use_alpha_beta)
            if new_score < value:
                value = new_score
                best_col = col
            if use_alpha_beta:
                beta = min(beta, value)
                if alpha >= beta:
                    break
        return best_col, value

def get_immediate_move(board, piece, rival):

    valid_cols = get_valid_locations(board)

    for col in valid_cols:
        temp_board = board.copy()
        drop_piece(temp_board, col, piece)
        if winning_move(temp_board, piece):
            return col

    for col in valid_cols:
        temp_board = board.copy()
        drop_piece(temp_board, col, rival)
        if winning_move(temp_board, rival):
            return col

    return None

def get_best_move(board, piece, rival, depth, use_alpha_beta):

    col_urgent = get_immediate_move(board, piece, rival)
    if col_urgent is not None:
        return col_urgent

    if piece == AI_1:

        col, _ = minimax(board, depth, -math.inf, math.inf, True, use_alpha_beta)
    else:

        col, _ = minimax(board, depth, -math.inf, math.inf, False, use_alpha_beta)
    return col


def play_game_human_vs_ai(use_alpha_beta):
    board = create_board()
    game_over = False
    turn = 0  

    depth_if_alpha = 6
    depth_if_noalpha = 4

    while not game_over:
        draw_board(board)

        if turn == 0:
            event = pygame.event.wait()
            if event.type == pygame.QUIT:
                pygame.quit()
                sys.exit()
            if event.type == pygame.MOUSEBUTTONDOWN:
                x_pos = event.pos[0]
                col = x_pos // SQUARESIZE
                if is_valid_location(board, col):
                    drop_piece(board, col, PLAYER)
                    if winning_move(board, PLAYER):
                        show_winner("¡Ganaste (Humano)!")
                        game_over = True
                    else:
                        if board_full(board):
                            show_winner("¡Empate!")
                            game_over = True
                    turn = 1
        else:
            time.sleep(0.3)
            depth = depth_if_alpha if use_alpha_beta else depth_if_noalpha

            col = get_best_move(board, AI_1, PLAYER, depth, use_alpha_beta)
            drop_piece(board, col, AI_1)
            if winning_move(board, AI_1):
                show_winner("La IA (Amarillo) ganó")
                game_over = True
            else:
                if board_full(board):
                    show_winner("¡Empate!")
                    game_over = True
            turn = 0

    if menu_selection("¿Jugar de nuevo?", ["Sí", "No"]) == 0:
        main()
    else:
        pygame.quit()
        sys.exit()

def play_game_ai_vs_ai(alpha_beta_ai):
    board = create_board()
    game_over = False

    depth_alpha = 6
    depth_noalpha = 4

    current_player = random.choice([AI_1, AI_2])

    first_move_done = False

    while not game_over:
        draw_board(board)
        time.sleep(0.3)

        use_poda = (alpha_beta_ai == current_player)
        depth = depth_alpha if use_poda else depth_noalpha

        if current_player == AI_1:
            rival = AI_2
        else:
            rival = AI_1

        if not first_move_done:
            valid_cols = get_valid_locations(board)
            col = random.choice(valid_cols)
            first_move_done = True
        else:
            col = get_best_move(board, current_player, rival, depth, use_poda)

        drop_piece(board, col, current_player)

        if winning_move(board, current_player):
            if current_player == AI_1:
                show_winner("IA_1 (Amarillo) ganó")
            else:
                show_winner("IA_2 (Naranja) ganó")
            game_over = True
        else:
            if board_full(board):
                show_winner("¡Empate!")
                game_over = True

        current_player = AI_2 if current_player == AI_1 else AI_1

    if menu_selection("¿Jugar de nuevo?", ["Sí", "No"]) == 0:
        main()
    else:
        pygame.quit()
        sys.exit()

def main():
    mode = menu_selection("Modo de juego", ["Humano vs IA", "IA vs IA"])
    if mode == 0:
        use_pruning = (menu_selection("¿Activar poda Alfa-Beta (IA)?", ["Sí", "No"]) == 0)
        play_game_human_vs_ai(use_pruning)
    else:
        who_uses_pruning = menu_selection(
            "¿Quién usa la poda Alfa-Beta?",
            ["AI_1 (Amarillo)", "AI_2 (Naranja)", "Ninguno"]
        )
        if who_uses_pruning == 0:
            alpha_beta_ai = AI_1
        elif who_uses_pruning == 1:
            alpha_beta_ai = AI_2
        else:
            alpha_beta_ai = None
        play_game_ai_vs_ai(alpha_beta_ai)

if __name__ == "__main__":
    main()
```
