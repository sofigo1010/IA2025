## 1. ¿Qué es el Temporal Difference Learning y en qué se diferencia de los métodos tradicionales de aprendizaje supervisado?

**Temporal Difference Learning (TD Learning):**  
Es un método de aprendizaje por refuerzo que permite actualizar de forma incremental la estimación de la función de valor a partir de la diferencia entre predicciones sucesivas. En lugar de esperar a la finalización de un episodio (como en los métodos Monte Carlo), TD Learning utiliza la retroalimentación inmediata obtenida en cada paso para ajustar las predicciones.

**Diferencias con el aprendizaje supervisado:**  
- **Etiquetado de datos:**  
  En el aprendizaje supervisado se requiere un conjunto de datos etiquetado (entrada-salida) para entrenar el modelo, mientras que en TD Learning no se cuentan con “etiquetas” explícitas; el aprendizaje se basa en la experiencia interactiva con el entorno.
- **Retroalimentación y actualización:**  
  En métodos supervisados, el ajuste se realiza comparando la predicción con una etiqueta verdadera. En TD Learning, la "etiqueta" se genera combinando la recompensa inmediata con la estimación del siguiente estado, utilizando lo que se conoce como **error de diferencia temporal**.

**Error de Diferencia Temporal (TD Error):**  
Se define como:
  
$$
\delta = r + \gamma V(s') - V(s)
$$

donde:  
- $`r`$ es la recompensa recibida,  
- $`\gamma`$ es el factor de descuento,  
- $`V(s)`$ es la estimación del valor del estado actual,  
- $`V(s')`$ es la estimación del valor del siguiente estado.

El TD error mide la discrepancia entre la predicción actual y la nueva estimación basada en la experiencia inmediata, y se utiliza para actualizar la función de valor de forma continua.

## 2. Decisiones en Juegos Simultáneos

En los **juegos simultáneos**, los jugadores toman decisiones al mismo tiempo, sin conocer las acciones que elegirán sus oponentes. Debido a esta incertidumbre, cada jugador debe basar su estrategia en las expectativas sobre el comportamiento de los demás.

**Ejemplo del mundo real:**  
- **Subastas cerradas:**  
  En una subasta cerrada, todos los participantes presentan sus ofertas simultáneamente sin conocer las ofertas de los demás. Cada postor debe estimar el valor del objeto y anticipar las pujas de sus competidores para decidir su propia oferta.

**Estrategias posibles:**  
- **Estrategia mixta:** Los jugadores pueden asignar probabilidades a diferentes acciones, mezclando sus estrategias para evitar ser predecibles.  
- **Modelado de creencias:** Cada jugador forma hipótesis sobre las posibles acciones de sus oponentes basándose en información histórica o en el contexto del problema.  
- **Optimización bajo incertidumbre:** Se busca maximizar la recompensa esperada, considerando los posibles resultados derivados de la incertidumbre en las acciones adversarias.

## 3. Juegos de Suma Cero vs. Juegos de No Suma Cero

**Juegos de Suma Cero:**  
- En estos juegos, la ganancia de un jugador es exactamente la pérdida del otro, de modo que el total de beneficios y pérdidas suma cero.  
- **Implicación en la toma de decisiones:**  
  Cada ventaja obtenida por un jugador implica una desventaja equivalente para su oponente, lo que fomenta estrategias estrictamente competitivas.

**Juegos de No Suma Cero:**  
- En estos juegos, el resultado total no es necesariamente cero, pudiendo existir situaciones en las que ambos jugadores ganen o pierdan simultáneamente.
- **Ejemplo:**  
  El **dilema del prisionero** es un claro ejemplo, donde la tentación individual de traicionar puede ser alta, pero la cooperación mutua produce un resultado mejor para ambos.
- **Consideraciones estratégicas:**  
  - **Cooperación vs. competencia:** Existe la posibilidad de estrategias cooperativas que beneficien a ambas partes.  
  - **Soluciones Pareto eficientes:** Los jugadores pueden buscar acuerdos o equilibrios en los que ninguna mejora unilateral sea posible sin perjudicar a la otra parte.  
  - **Equilibrio de Nash en contextos cooperativos:** La estabilidad del equilibrio puede requerir compromisos y mecanismos para fomentar la confianza entre los jugadores.

## 4. Equilibrio de Nash en Juegos Simultáneos

El **equilibrio de Nash** se aplica en juegos simultáneos como una situación en la que cada jugador elige una estrategia de modo que, dado el conjunto de estrategias de los demás, ningún jugador tiene incentivos para cambiar su elección de manera unilateral.

- **Estabilidad:**  
  En el equilibrio de Nash, cualquier cambio individual sin coordinación llevaría a un peor resultado para el jugador que se desvía.
- **Aplicación en juegos simultáneos:**  
  Dado que los jugadores seleccionan sus estrategias sin conocer las elecciones de los demás, el equilibrio de Nash es una predicción razonable de comportamiento en escenarios con información incompleta.
- **Ejemplo:**  
  En el dilema del prisionero, si ambos jugadores adoptan la estrategia que representa el equilibrio de Nash (generalmente la traición en el contexto clásico), ninguno ganará en mejorar su resultado cambiando de forma unilateral.

## 5. Aplicación del Temporal Difference Learning en Procesos de Toma de Decisiones Dinámicos

**Modelado y Optimización en Entornos Dinámicos:**  
TD Learning se utiliza para optimizar procesos de toma de decisiones en entornos donde las condiciones cambian rápidamente. Al actualizar continuamente las estimaciones de valor en función de la experiencia inmediata, el método se adapta dinámicamente a nuevos escenarios y recompensas.

**Equilibrio entre Exploración y Explotación:**  
- **Explotación:** Se aprovecha el conocimiento actual para maximizar la recompensa inmediata.  
- **Exploración:** Se prueban nuevas acciones para descubrir información que pueda conducir a mejores recompensas en el futuro.
- **Mecanismos comunes:**  
  - Estrategia **$\varepsilon$-greedy**, en la que con probabilidad $`\varepsilon`$ se elige una acción aleatoria (exploración) y con probabilidad $`1-\varepsilon`$ se selecciona la acción óptima (explotación).  
  - Otros enfoques incluyen métodos basados en la incertidumbre, como el Upper Confidence Bound (UCB).

**Desafíos en la Implementación Práctica:**  
- **Convergencia:** Asegurar que las actualizaciones basadas en el TD error conduzcan a una función de valor estable.  
- **Selección del parámetro de exploración:** Determinar la tasa de exploración adecuada para evitar tanto la exploración excesiva como la explotación prematura.  
- **Aproximación de funciones:** En entornos con un gran número de estados o estados continuos, se utilizan métodos de aproximación (como redes neuronales), lo que puede introducir problemas de estabilidad y convergencia.  
- **Equilibrio a largo plazo:** Manejar la discrepancia entre la recompensa inmediata y la recompensa futura, crucial en entornos complejos y no estacionarios.




# Laboratorio 7 – Connect Four con TD Learning

En este notebook se implementa un agente para Connect Four basado en Temporal Difference Learning (TD Learning).  
El agente aprende a partir de sus propias experiencias actualizando una función de valor basada en el error de diferencia temporal.  

Los pasos principales son:
1. **Representación del Estado:**  
   Se utiliza el tablero (una matriz de NumPy) y se codifica en un vector (o tupla) que sirva de clave para la tabla Q.
2. **Espacio de Acción:**  
   Las acciones corresponden a las columnas en las que se puede dejar caer una pieza, es decir, las posiciones válidas del tablero.
3. **Algoritmo TD (Q-learning):**  
   La actualización de la función de valor se realiza usando la regla de Q-learning:  
   $$ Q(s,a) \leftarrow Q(s,a) + \alpha \left[ r + \gamma \max_{a'} Q(s',a') - Q(s,a) \right] $$
4. **Definición de Recompensas:**  
   - Victoria del agente: recompensa +1  
   - Derrota: recompensa -1  
   - Empate: 0  
   También se pueden definir recompensas intermedias para incentivar comportamientos favorables.
5. **Estrategia de Exploración:**  
   Se emplea una política $\varepsilon$-greedy para equilibrar exploración y explotación.
6. **Ciclo de Entrenamiento y Evaluación:**  
   El agente TD se enfrenta a dos tipos de agentes Minimax (uno sin poda y otro con poda alfa-beta) en 50 juegos cada uno, totalizando 150 juegos.  
   Luego se grafica la cantidad de victorias de cada agente.
